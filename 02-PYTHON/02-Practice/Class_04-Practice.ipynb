```python
def pipeline_event_logger(*args, **kwargs):
    for item in args:
        print(item.upper())
    for key, value in kwargs.items():
        if key == "" or value == "":
            print("Warning! Something went wrong.")
        else:
            print(key, ":", value)
pipeline_event_logger(
    "high_memory_usage", "retry_successfull", "critical","retry_exceeded","s3_timeout",
    pipeline_name="sales_etl",                
    status="SUCCESS",                         
    rows_processed=1500,                     
    source_db="PostgreSQL",                  
    execution_time_sec=12.4
)
```

```python
def calculate_cart_total(*args, **kwargs):
    subtotal = sum(args)
    print(f"Your Subtotal is {subtotal} pkr")

    #for key, value in kwargs.items():
    discount_amount = subtotal * (kwargs.get('discount', 0) / 100)
    total = subtotal - discount_amount
    tax_amount = total * (kwargs.get('tax', 0) / 100) 
    print(f"Discount granted: {discount_amount} pkr")
    print(f"Tax applied: {round(tax_amount, 2)} pkr\n")

    print(f"Your Total after discount is: {total} pkr")

    new_total = total + tax_amount
    print(f"Your Total after applying tax is: {new_total} pkr")

calculate_cart_total(
    100, 500, 30,
    discount = 20,
    tax = 5,
    coupon = "FLAT10",
    delivery = 150
)
```
