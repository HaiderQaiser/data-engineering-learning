1.2 — You have a storage budget of 500 GB. A pipeline writes 37 GB per day. Calculate how many full days until the storage runs out, and how many GB remain on the last day. Use only // and %.

```python
storage = 500
per_day = 37

print("The days until the storage runs out are: ", storage // per_day)
print("Remaining GBs: ", storage % per_day)
```

1.4 — Create a list of five scores. Print the highest (use max()), lowest (use min()), and the total (use sum()). Then append one more score and print the new average.

```python
score_list = [25,50,99,70,80]
print(max(score_list))
print(min(score_list))
print(sum(score_list))

score_list.append(100)
print("Average: ", round(sum(score_list) / len(score_list),2))
```

1.7 — A load reports rows_loaded, rows_expected, and errors. Write a program that prints "Publish" when at least 99% of expected rows loaded and there are no errors, "Review" when the load is between 95% and 99% or there are errors, and "Stop" when fewer than 95% loaded. Test it with three different sets of values, including one that shows why the order of your branches matters.

```python
# Test Case 1: Publish

rows_loaded = 995
rows_expected = 1000
errors = 0

load_percent = (rows_loaded / rows_expected) * 100

if load_percent >= 99 and errors == 0:
    print("Publish")
elif load_percent >= 95 or errors > 0:
    print("Review")
else:
    print("Stop")


# Test Case 2: Review

rows_loaded = 960
rows_expected = 1000
errors = 0

load_percent = (rows_loaded / rows_expected) * 100

if load_percent >= 99 and errors == 0:
    print("Publish")
elif load_percent >= 95 or errors > 0:
    print("Review")
else:
    print("Stop")


# Test Case 3: Stop

rows_loaded = 900
rows_expected = 1000
errors = 0

load_percent = (rows_loaded / rows_expected) * 100

if load_percent >= 99 and errors == 0:
    print("Publish")
elif load_percent >= 95 or errors > 0:
    print("Review")
else:
    print("Stop")
```
