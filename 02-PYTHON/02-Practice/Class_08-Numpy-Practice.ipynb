6.1 — Create an array of 20 random integers between 50 and 150. Print its shape, dtype, min, max, mean, and standard deviation. Then sort it and print the sorted version.

```python
import numpy as np

# 1. Create an array of 20 random integers between 50 and 150
ran_arr = np.random.randint(50, 151, size=20)

# Print array
print("Original Array:\n", ran_arr)
print("-" * 30)

# Shape & dtype
print("Shape:", ran_arr.shape)
print("Data Type:", ran_arr.dtype)

# Aggregations (Min, Max, Mean, Std)
print("Min:", ran_arr.min())
print("Max:", ran_arr.max())
print("Mean:", ran_arr.mean())
print("Standard Deviation:", np.std(ran_arr)) 

# Sorting & printing sorted version
sorted_arr = np.sort(ran_arr)
print("Sorted Array:\n", sorted_arr)
```

6.2 — Create two 1D arrays of length 6 representing monthly revenue and monthly costs. Calculate profit for each month, total annual profit, and the month with the highest profit using np.argmax(). Print all results.

```python
import numpy as np

arr1 = np.array([100,300,100,500,400,90])
arr2 = np.array([30,90,80,50,100,45])

monthly_profit = arr1 - arr2
annual_profit = np.sum(monthly_profit)

print(f"Profit: {monthly_profit}")
print("Annual Profit: ", annual_profit)

month_with_highest_profit = monthly_profit.argmax()

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
best_month = months[month_with_highest_profit] 

print(f"Month with the highest profit: {best_month} \nTotal Profit in this month: {np.max(monthly_profit)}")
```

6.5 — Create an array of 15 exam scores between 40 and 100. Use boolean indexing to extract all scores above 70. Use np.where() to label each score "Distinction" (≥85), "Pass" (≥50), or "Fail" (below 50) — using nested np.where() calls.

```python
scores = np.array([42, 58, 64, 71, 88, 95, 49, 77, 83, 91, 60, 74, 85, 99, 53])

results = np.where(scores >= 85 , "Distincion", 
                   np.where(scores >= 50, "Pass", "Fail") )
print(results)
```

6.6 — Create a 3×4 array of sales figures. Compute: the total sales for each row (use axis=1), the average for each column (use axis=0), and the single highest value across the entire array with its row and column position using np.argmax() and np.unravel_index().

```python
import numpy as np

sales = np.arange(100, 1300, 100).reshape(3, 4)
print(sales)
max_sales = sales.max()
print(max_sales)

flat_max_idx = sales.argmax()
print(flat_max_idx)

row_idx, col_idx = np.unravel_index(flat_max_idx, sales.shape)  # Translate to 2D

print(f"Highest sales value: {max_sales}")
print(f"Position of highest sales: Row Index {row_idx}, Column Index {col_idx}")
```
