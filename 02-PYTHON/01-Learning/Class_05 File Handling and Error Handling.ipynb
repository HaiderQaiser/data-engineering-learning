```python
import csv
with open("students.csv", "w", newline="") as f:
    file = csv.writer(f)
    file.writerow(["Name", "Class"])
    file.writerows([
                    ["Haider", 12],
                    ["Muaz", 11]
                    ])
```

```python
import csv
with open("students.csv", "r") as file:
    data = csv.reader(file)
    #print(data)
    data_in_list = list(data)
    #print(data_in_list)

    data_in_list[0].append("marks")
    data_in_list[1].append(90)
    data_in_list[2].append(40)
    #print(data_in_list)
    #data.writerows(data_in_list)
    with open("passed_students.csv", "a", newline="") as new_file:
        updated_file = csv.writer(new_file)
        #file mn header pehle hi likh dia 
        updated_file.writerow(data_in_list[0])

        for i in range(1, len(data_in_list)):
            if int(data_in_list[i][2]) > 60:
                updated_file.writerow(data_in_list[i])
```
