## Abstraction

```python
# Abstraction
class Employee:
    def __init__(self, name):
        self.name = name
    def calculate_salary(self):
        pass
class FullTimeEmployee(Employee):
    def __init__(self,name, salary):
        super().__init__(name)
        self.name = name
        self.salary = salary
    def calculate_salary(self):
        print(f"{self.name}'s Full Time Salary is {self.salary}") 
class PartTimeEmployee(Employee):
    def __init__(self, name,hours_worked, hourly_rate):
        super().__init__(name)
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate
    def calculate_salary(self):
        print(f"{self.name}'s Part time salary is {self.hours_worked * self.hourly_rate}")
class ContractEmployee(Employee):
    def __init__(self, name, contract_amount):
        super().__init__(name)
        self.name = name
        self.contract_amount = contract_amount
    def calculate_salary(self):
        print(f"{self.name}'s Contract Salary is {self.contract_amount}") 
```

```python
obj_1 = PartTimeEmployee("Haider", 8, 500)
obj_1.calculate_salary()
```

```python
obj_2 = FullTimeEmployee("Muaz", 80000)
obj_2.calculate_salary()
```

```python
obj_3 = ContractEmployee("Ali", 5000)
obj_3.calculate_salary()
```

## Polymorphism

```python
# Polymorphism
class Notification:
    def __init__(self, notification, message):
        self.notification = notification
        self.message = message

    def send_notification(notification, message):
        print(f"{notification} appeared! The message is: '{message}'")

class EmailNotification(Notification):
    def __init__(self, notification, message):
        super().__init__(notification, message)

    def send_notification(self):
        print(f"{self.notification} appeared. The message is: {self.message}")

class SMSNotification(Notification):
    def __init__(self, notification, message):
        super().__init__(notification, message)

    def send_notification(self):
        print(f"{self.notification} appeared. The message is: {self.message}")
        

class PushNotfication(Notification):
    def __init__(self, notification, message):
        super().__init__(notification, message)
    
    def send_notification(self):
        print(f"{self.notification} appeared. The message is: {self.message}")
```

```python
not_1 = EmailNotification("Duolingo", "Do your lesson on duolingo!")
not_1.send_notification()
```

```python
not_2 = SMSNotification("Instagram", "__the_haider__ recently added to their story")
not_2.send_notification()
```

## Encapsulation

```python
class saylani:
    def __init__(self, name, email, cnic):
        self.name = name
        self._email = email
        self.__cnic = cnic
    def public_method(self):
        print("This is a Public method")

    def _protected_method(self):
        print("This is a Protected method")

    def __private_method(self):
        print("This is a Private method")
```

```python
obj_1 = saylani("Haider", "xyz@gmail.com", "42201-0000-00")
obj_1.public_method()
```
