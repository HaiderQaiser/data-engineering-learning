## FOR LOOP

```python
bio = {
    "name" : "Haider",
    "age" : 18
}
for key, value in enumerate(bio):
    print(f"key '{key}' have value '{value}'")
```

```python
nums = [100,4,200,1,3,2,5,8,7,6]
nums.sort()
#print(l)

for i in range(min(nums) - 1, max(nums)):
    if nums[i] == i+1:
        print(nums[i])8.
    else:
        break
```

## WHILE LOOP

```python
nums = [1,2,20,4,5,6,7,8]
even = []
odd = []

i = nums[0]
while i < len(nums):
    if i % 2 == 0:
        even.append(i)
        i += 1
    else:
        odd.append(i)
        i += 1
print(even)
print(odd)
```

## FUNCTIONS

```python
def bio_data(name, age, weight, height):
    print(f" Your name is: {name} \n Your age is: {age} \n Your weight is: {weight} \n Your height is: {height}")

bio_data("Haider", 18, 50, 5.6)
```

```python
def make_pizza(size, *toppings):
    print(f"\nMaking a {size}-inch pizza with the following toppings:")
    # toppings yahan aik tuple ban chuka hai
    for topping in toppings:
        print(f"- {topping}")

# Aap jitne chahein toppings pass kar sakte hain
make_pizza(12, "pepperoni", "mushrooms", "extra cheese")
make_pizza(8, "green peppers")
```

```python
def build_profile(first, last, **user_info):
    # user_info yahan aik dictionary ban chuka hai
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info

# Aap extra information key=value ki surat mein de sakte hain
user_profile = build_profile("Haider", "Ali", age=18, city="Karachi")
print(user_profile)
# Output: {'age': 18, 'city': 'Karachi', 'first_name': 'Haider', 'last_name': 'Ali'}
```
