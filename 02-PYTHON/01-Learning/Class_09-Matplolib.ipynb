```python
import matplotlib.pyplot as plt
from datetime import datetime,timedelta
import numpy as np
```

```python
year = [2020,2021,2022,2023,2024,2025]
sales = [100,105,120,130,270,70,]

plt.title("Year Wise Sales")
plt.plot(year, sales)
plt.xlabel("Years")
plt.ylabel("Sales")
plt.show()
plt.close()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [100,105,120,130,270,70,]

plt.title("Year Wise Sales")
plt.bar(months, sales)
plt.xlabel("Years")
plt.ylabel("Sales")
plt.show()
plt.close()
```

```python
start_date = datetime(2024,1,1)
end_date = datetime(2024,12,31)
num_days = (end_date - start_date).days
print(start_date, num_days)
```

```python
date_list = []
for i in range(num_days):
    date_list.append(start_date.date())
    start_date += timedelta(days=1)
date_list
```

```python
stock_price_list = (np.random.randn(num_days)).tolist()
```

```python
np.random.seed(0) # jitni baar bhi run krengy random values same aengi
stock_price_list = (np.random.randn(num_days).cumsum()*50).tolist() # cumsum is used to get cumulative sum
stock_price_list
```

```python
plt.Figure(figsize=(15,6))
plt.plot(date_list, stock_price_list, color='red', label='price')
plt.title("STOCK PRICE OF ONE YEAR")
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.legend() 
plt.grid(True)
plt.tight_layout()
plt.savefig("Stock_price_report.png", dpi=300)
plt.show()
plt.close()
```

```python
plt.figure(figsize=(15,6))
plt.plot(date_list, stock_price_list, color='green',marker='.',label='price')
plt.title("STOCK PRICE OF ONE YEAR")
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.legend() # used to access price label 
plt.grid(True)
plt.tight_layout()
plt.show()
plt.close()
```

```python
plt.figure(figsize=(15,6))
plt.plot(date_list, stock_price_list, color='green',linestyle=':',label='price')
plt.title("STOCK PRICE OF ONE YEAR")
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.legend() # used to access price label 
plt.grid(True)
plt.tight_layout()
plt.show()
plt.close()
```

```python
stock_price_list2 = (np.random.randn(num_days).cumsum()*20).tolist()
```

```python
plt.figure(figsize=(15,6))
plt.plot(date_list, stock_price_list, color='green',linestyle='--', label='AAPL') # making two graphs this time
plt.plot(date_list, stock_price_list2, color='red',linestyle='--',label='MSFT')
plt.title("STOCK PRICE OF ONE YEAR")
plt.xlabel("Date")
plt.ylabel("Stock Price")
plt.legend() 
plt.grid(True)
plt.tight_layout() 
plt.show()
plt.close()
```
