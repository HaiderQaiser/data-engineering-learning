## INHERITENCE

```python
class Father:
    def __init__(self, f_name, blood_group):
        self.f_name = f_name
        self.blood_group = blood_group

    def Fav_food(self, fav_food ):
        print(f"{self.f_name},s favourite food is {fav_food}")

class Mother:
    def __init__(self, m_name, blood_group):
        self.f_name = m_name
        self.blood_group = blood_group

    def Fav_hobby(self, fav_hobby ):
        print(f"{self.f_name},s favourite food is {fav_hobby}")

# Child Class
class Child(Father, Mother):
    def __init__(self,f_name, m_name, name, age, blood_group):
        super().__init__(f_name, blood_group)
        super().__init__(m_name, blood_group)
        self.name = name
        self.age = age
        self.blood_group = blood_group

father = Father("Muhammad", "A+")
father.Fav_food("Biryani")
mother = Mother("Sara", "B+")
mother.Fav_hobby("Cooking")
child_01 = Child("Ali", 14, "AB+")

child_01.Fav_food()
child_01.Fav_hobby()
```

```python
# Correct Code
class Father:
    def __init__(self, f_name, blood_group):
        self.f_name = f_name
        self.f_blood_group = blood_group

    def fav_food(self, food):
        print(f"{self.f_name}'s favourite food is {food}")


class Mother:
    def __init__(self, m_name, blood_group):
        self.m_name = m_name
        self.m_blood_group = blood_group

    def fav_hobby(self, hobby):
        print(f"{self.m_name}'s favourite hobby is {hobby}")


# Child Class (Multiple Inheritance)
class Child(Father, Mother):
    def __init__(self, f_name, m_name, name, age, blood_group):
        # Multiple Inheritance mein explicit parent constructors call kar rahe hain
        Father.__init__(self, f_name, blood_group)
        Mother.__init__(self, m_name, blood_group)
        
        self.name = name
        self.age = age
        self.blood_group = blood_group


# Object Creation
father = Father("Muhammad", "A+")
father.fav_food("Biryani")

mother = Mother("Sara", "B+")
mother.fav_hobby("Cooking")

# Poore 5 arguments pass kar rahe hain
child_01 = Child("Muhammad", "Sara", "Ali", 14, "AB+")

# Required arguments pass karke method call
child_01.fav_food("Pulao")
child_01.fav_hobby("Cricket")
```
